```python
from scipy import stats
```

```python
print(stats.describe([60, 70, 80, 90, 100])) #分析一組資料，例如資料數量、平均值、變異程度、最大值和最小值等
```

```python
# 計算平均數
scores = [60, 70, 80, 90, 100]

print(stats.tmean(scores))
```

```python
# 計算眾數（出現最多次的數值）
data = [1, 2, 2, 3, 3, 3, 4]
result = stats.mode(data)

print(result.mode)
print(result.count)
```

```python
# 常態分布（Normal Distribution）
property = stats.norm.cdf(80, loc=70, scale=10)
print(property)
```

```python
# 假設檢定（Hypothesis Testing）

scores = [72, 78, 80, 74, 77, 81, 69, 76]

result = stats.ttest_1samp(scores, popmean=75)

print("t 統計量:", result.statistic)
print("P值 :", result.pvalue)
```

```python
hours = [1, 2, 3, 4, 5]
scores = [50, 55, 65, 75, 85]

result = stats.pearsonr(hours, scores)

print("相關係數：", result.statistic)
print("p 值：", result.pvalue)
```

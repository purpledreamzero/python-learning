```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, 'o-g')
plt.show()
```

```python
# Matplotlib 折線圖
plt.plot([1, 2, 3, 4], [10, 11, 15, 10])
plt.show()
```

```python
# Matplotlib散點圖
plt.scatter([1, 2, 3, 4], [10, 20, 12, 18])
plt.show()
```

```python
# Matplotlib餅圖
plt.pie([20, 30, 50], labels=['A', 'B', 'C'])
plt.show()
```

```python
# Matplotlib 柱狀圖
plt.bar(['A', 'B', 'C'], [20, 25, 10], color='black') #color設定顏色
plt.show()
```

```python
# 圖表的美化
plt.title('test')
plt.xlabel('XXXX')
plt.ylabel('YYYY')
plt.grid()      #設置網格
```

```python
# 保存圖片
plt.savefig('aaa.png')
plt.show()
```

```python
# Matplotlib 多圖顯示
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3,], [4, 5, 3])
plt.subplot(1, 2, 2)
plt.bar(['A', 'B', 'C'], [100, 200, 140])

plt.show()
```

```python
# 結合使用 pandas
df = pd.DataFrame({'x':[1, 2, 3], 'y':[4, 5, 6]})
plt.plot(df['x'], df['y'])
plt.show()
```

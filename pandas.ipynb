```python
import yfinance as yf
import pandas as pd

def get_stock(ticker='2330.TW'):
    # 1. 下載股票資料
    df_2330 = yf.download(
        ticker,
        period='2y',
        auto_adjust=True,
        progress=False
    )

    # 2. 檢查是否有下載到資料
    if df_2330.empty:
        raise ValueError(f'無法取得股票資料：{ticker}')

    # 3. 計算移動平均線
    close = df_2330['Close']

    df_2330['MA20'] = close.rolling(20).mean()
    df_2330['MA60'] = close.rolling(60).mean()
    df_2330['MA120'] = close.rolling(120).mean()
    df_2330['MA240'] = close.rolling(240).mean()

    # 4. 回傳完整資料
    return df_2330


# 呼叫函式
df_2330 = get_stock()

print(df_2330.tail())
```

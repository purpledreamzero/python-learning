```python
import pandas as pd

#1. 載入銷售資料
df = pd.read_csv('sales.csv')

#2. 檢查資料基本狀況
print(df.shape) # 資料有幾列、幾欄
print(df.dtypes) # 每個欄位的資料型態
print(df.isna().sum()) # 每個欄位有多少缺失值

#3. 計算銷售金額
df["銷售金額"] = df["單價"] * df["數量"]

#4. 按商品總銷售金額
summary = (
    df.groupby("商品", as_index=False)["銷售金額"]
    .sum()
    .sort_values("銷售金額", ascending=False)
)

#5. 輸出分析結果
summary.to_csv(
    'sales_summary.csv',
    index=False,
    encoding='utf-8-sig'
)
```

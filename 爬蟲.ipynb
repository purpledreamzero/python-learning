```python

import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. 指定目標網站
url = "https://books.toscrape.com/"

# 2. 取得網頁
response = requests.get(url, timeout=15)
response.raise_for_status()

# 3. 解析 HTML
soup = BeautifulSoup(response.text, "html.parser")

# 4. 找出每一本書
books = soup.select("article.product_pod")

# 5. 擷取書名、價格、庫存狀態
data = []

for book in books:
    title = book.h3.a["title"]
    price = book.select_one(".price_color").get_text(strip=True)
    stock = book.select_one(".availability").get_text(strip=True)

    data.append({
        "書名": title,
        "價格": price,
        "庫存狀態": stock
    })

# 6. 整理成表格
df = pd.DataFrame(data)

print("抓取筆數：", len(df))
display(df)
```

```python

import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. 設定要抓取的兩個頁面
urls = [
    "https://books.toscrape.com/",
    "https://books.toscrape.com/catalogue/page-2.html"
]

# 2. 建立一份共同的資料清單
data = []

# 3. 逐頁抓取
for url in urls:
    response = requests.get(url, timeout=15)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.select("article.product_pod")

    # 4. 逐本書擷取資料
    for book in books:
        title = book.h3.a["title"]
        price = book.select_one(".price_color").get_text(strip=True)
        stock = book.select_one(".availability").get_text(strip=True)

        data.append({
            "書名": title,
            "價格": price,
            "庫存狀態": stock
        })

# 5. 合併成一份表格
df = pd.DataFrame(data)

print("總抓取筆數：", len(df))
display(df)
```

Requests GET、Response、狀態碼與錯誤處理

1.發送 GET 請求

使用 requests.get() 向網站發送 HTTP GET 請求，取得伺服器回傳的 Response 物件。

import requests

url = "https://books.toscrape.com/"
response = requests.get(url)

2.查看回應內容

print(response)
print(response.status_code)
print(response.text[:200])

response：HTTP 回應物件。

response.status_code：HTTP 狀態碼。

response.text：回應的文字內容，通常包含 HTML。

3.設定 timeout

response = requests.get(url, timeout=3)

設定連線與讀取逾時限制，避免程式無限期等待。逾時可能引發 requests.exceptions.Timeout。

4.使用 raise_for_status()

response.raise_for_status()

當回應狀態碼為 HTTP 4xx 或 5xx 時，會拋出 HTTP 錯誤例外。

5.使用 try / except 處理錯誤

import requests

url = "https://books.toscrape.com/"

try:
    response = requests.get(url, timeout=3)
    response.raise_for_status()
    print("請求成功!")

except requests.exceptions.RequestException as error:
    print("請求失敗:", error)

6.紀錄

正常頁面測試：HTTP 200，請求成功。

不存在的頁面測試：HTTP 404，成功進入錯誤處理區塊。
